```python
from pathlib import Path
import json
import subprocess
import marimo as mo
PROJECT = Path('/marimo/world_models_restored_20261103')
RECOVERY = PROJECT / 'runs/recovery-20261103'
receipt = json.loads((RECOVERY / 'RESTORATION.json').read_text())
assert len(receipt['reports']) == 12
print('Source:', receipt['source_commit'])
print('Verified inventory entries:', sum(a['files_verified'] for a in receipt['archives']))
print('Preserved diagnostic reports:', len(receipt['reports']))
print('Original data:', receipt['data'])
print('Python:', PROJECT / '.venv/bin/python')
print('Completed study experiments rerun:', receipt['experiments_launched'])
print(subprocess.check_output(['nvidia-smi', '--query-gpu=name,memory.used,utilization.gpu', '--format=csv,noheader'], text=True))
preflight = PROJECT / 'runs/recovery-preflight.json'
print('Models passing native GPU preflight:', list(json.loads(preflight.read_text())['models']) if preflight.exists() else 'Pending; inspect preflight-recovery.log')
```

```python
panels = [mo.md('## Restored manuscript and recovery receipt') ]
for artifact in [PROJECT / 'PAPER_DRAFT.md', PROJECT / 'FOLLOWUP_RESULTS_20261102.md', PROJECT / 'runs/molab-20261102-retrieved/figures/followup_summary.pdf', RECOVERY / 'RESTORATION.json', PROJECT / 'runs/recovery-preflight.json']:
    if artifact.is_file():
        panels.append(mo.download(data=artifact.read_bytes(), filename=artifact.name, label='Download ' + artifact.name))
mo.vstack(panels)
```

```python
BATCH = PROJECT / 'runs/paper-studies-20261103-01'
status_path = BATCH / 'status.json'
if status_path.is_file():
    status = json.loads(status_path.read_text())
    print('Batch:', status['state'], 'updated:', status['updated_utc'])
    print('Source:', status['source_commit'])
    for name, item in status['stages'].items():
        print('\n', name, item['state'], 'PID:', item.get('pid'), 'exit:', item.get('exit_code'))
        print('Progress:', item.get('line', 'Not started'))
        log = BATCH / (name + '.log')
        if log.is_file():
            with log.open('rb') as stream:
                stream.seek(max(0, log.stat().st_size - 2500))
                print(stream.read().decode('utf-8', errors='replace'))
        if item.get('archive'):
            print('Verified archive:', item['archive'])
else:
    print('Tracked batch has not started. This read-only monitor never launches it.')
```
